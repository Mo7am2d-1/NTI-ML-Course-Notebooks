## 11) فحص الـ Outliers

```python
data.boxplot()  # عرض بصري سريع
```

```python
Q1 = data['Newspaper'].quantile(0.25)
Q3 = data['Newspaper'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# عرض الـ outliers الأول قبل ما نشيلهم
outliers = data[(data['Newspaper'] < lower_bound) | (data['Newspaper'] > upper_bound)]
print(outliers)
```

```python
# حذف الـ outliers
data = data[(data['Newspaper'] >= lower_bound) & (data['Newspaper'] <= upper_bound)]
```

## 12) تجهيز X و Y

```python
X = data[["TV", "Radio", "Newspaper"]]
X
```

```python
Y = data["Sales"]
Y
```

## 13) تقسيم البيانات (Train / Test Split)

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.2,      # 20% تست، 80% تدريب
    random_state=42     # يضمن نفس التقسيم في كل مرة تشغل الكود
)

print(f"X_train: {X_train.shape}, X_test: {X_test.shape}")
print(f"y_train: {y_train.shape}, y_test: {y_test.shape}")
```

## 14) توحيد المقياس (Scaling)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)   # fit + transform على التدريب فقط
X_test_scaled = scaler.transform(X_test)          # transform بس على التست (من غير fit)
```

## 15) تدريب الموديل (Linear Regression)

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
```

## 16) تقييم الموديل بالـ MSE

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
print(f"MSE: {mse:.2f}")
```
