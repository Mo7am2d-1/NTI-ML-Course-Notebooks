```python
df.drop('name' , axis=1 , inplace=True)
```

```python
df = pd.get_dummies(
    df,
    columns=['fuel', 'seller_type', 'transmission', 'owner'],
    drop_first=True,
    dtype=int
)
```

```python
np.max(df)
```

```python
np.min(df)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
scale = ['year', 'km_driven']
df[scale] = scaler.fit_transform(df[scale])
```

```python
df.head()
```

```python
q1 = df.quantile(0.25)
q3 = df.quantile(0.75)

iqr = q3 - q1

lower = q1 - (1.5 * iqr)
upper = q3 + (1.5 * iqr)
```

```python
((df < lower) | (df > upper)).sum()
outliers = df[((df < lower) | (df > upper)).any(axis=1)]
```

```python
print(outliers)
```

```python
df = df[~((df < lower) | (df > upper)).any(axis=1)]
```

```python
print=(outliers)
```

```python
sns.pairplot(df)
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=65)
plt.show()
```

```python
x=df[['year', 'km_driven', 'fuel_Diesel', 'fuel_Electric',
       'fuel_LPG', 'fuel_Petrol', 'seller_type_Individual',
       'seller_type_Trustmark Dealer', 'transmission_Manual',
       'owner_Fourth & Above Owner', 'owner_Second Owner',
       'owner_Test Drive Car', 'owner_Third Owner']].values
y=df['selling_price'].values  
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=.34,    
    random_state=66 )
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test, y_pred)
print(mse)
```
