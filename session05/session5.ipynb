```python
import numpy as np              # numerical computations
import pandas as pd             # data manipulation and analysis
import matplotlib.pyplot as plt # plotting and visualization
import statsmodels.api as sm    # statistical modeling
import matplotlib.image as img
import seaborn as sns
import plotly.express as px
import sklearn as sk
import sys
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df=pd.read_csv(r"advertising11.csv")
```

```python
df. head()
```

```python
x = df[["TV", "Radio", "Newspaper"]].values
y = df["Sales"].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,      # 20% for testing, 80% for training
    random_state=42     # ensures reproducibility - same split every time you run it
)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test, y_pred)
print(mse)
```

```python
df=pd.read_csv("auto-mpg[1].csv")
```

```python
from sklearn.preprocessing import OrdinalEncoder
             
```

```python
df.drop("car name", axis=1, inplace=True)
```

```python
df = df[df["horsepower"] != "?"]
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
q1 = df["displacement"].quantile(0.25)
q3 = df["displacement"].quantile(0.75)

IQR = q3 - q1

lower = q1 - 1.5 * IQR
upper = q3 + 1.5 * IQR

outliers = df[(df["displacement"] < lower) | (df["displacement"] > upper)]

outliers
```

```python
x = df[['cylinders', 'cylinders', 'horsepower','weight','acceleration','model year','origin']].values
y = df['mpg'].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,    
    random_state=42 )
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test, y_pred)
print(mse)
```
