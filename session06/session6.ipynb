```python
df=pd.read_csv(r"LR Data.csv")
```

```python
df=df.drop(columns=['Timestamp'])
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
cols=['City','Country','Ad Topic Line']
for col in cols:
    encoder= LabelEncoder()
    df[col]= encoder.fit_transform(df[col])
```

```python
x=df.drop(['Clicked on Ad'], axis=1)
y=df['Clicked on Ad']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,shuffle=True,random_state=42)
```

```python
from sklearn.feature_selection import mutual_info_classif
```

```python
info_gain= mutual_info_classif(x_train, y_train, random_state=42)
```

```python
for feature, gain in enumerate(info_gain):
    print(f'Info Gain for Feature {feature} =', gain)
```

```python
top_k = np.argsort(info_gain)[::-1][:5]
print(top_k)
```

```python
x_train2=x_train.iloc[:,top_k]
x_test2=x_test.iloc[:,top_k]
```

```python
x_train2[['Area Income']] = scaler.fit_transform(x_train2[['Area Income']])
x_test2[['Area Income']] = scaler.transform(x_test2[['Area Income']])
```

```python
model.fit(x_train2,y_train)
y_pred1= model.predict(x_test2)
```

```python
accuracy_score(y_test,y_pred1)
```
