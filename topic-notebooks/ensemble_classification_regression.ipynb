# classification

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

#models
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from xgboost import XGBClassifier
```

```python
df = pd.read_csv('/content/customer_data.csv')
df.head()
```

```python
df=df.drop(['id'],axis=1)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['fea_2'].fillna(df['fea_2'].mean(),inplace=True)
```

```python
df.describe()
```

```python
x=df.drop(['label'],axis=1)
y=df['label']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,shuffle=True,random_state=42)
```

```python
scaler = StandardScaler()
```

```python
x_train[['fea_2','fea_4','fea_10','fea_11']]=scaler.fit_transform(x_train[['fea_2','fea_4','fea_10','fea_11']])
```

```python
x_test[['fea_2','fea_4','fea_10','fea_11']]=scaler.transform(x_test[['fea_2','fea_4','fea_10','fea_11']])
```

```python
models={
    'KNN':KNeighborsClassifier(),
    'Logistic Regression':LogisticRegression(),
    'SVC':SVC(),
    'Decision Tree':DecisionTreeClassifier(),
    'Random Forest':RandomForestClassifier(),
    'AdaBoost':AdaBoostClassifier(),
    'XGBoost':XGBClassifier()
}
```

```python
def evaluate_model(actual,pred):
  report = classification_report(actual,pred)
  score = accuracy_score(actual,pred)
  return report,score
```

```python
import warnings
warnings.filterwarnings('ignore')
for i in range(len(list(models))):
  model=list(models.values())[i]  # call

  model.fit(x_train,y_train)      # fit

  y_pred=model.predict(x_test)    # predict


  report,score=evaluate_model(y_test,y_pred)
  print(list(models.keys())[i])
  print('performance:')
  print(score)
  print(report)
```

#Regression

```python
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
```

```python
df= pd.read_csv('/content/StudentsPerformance.csv')
```

```python
df.info()
```

```python
x=df.drop(['math score'],axis=1)
y=df['math score']
```

```python
encodr=LabelEncoder()
```

```python
x['gender']=encodr.fit_transform(x['gender'])
x['race/ethnicity']=encodr.fit_transform(x['race/ethnicity'])
x['lunch']=encodr.fit_transform(x['lunch'])
x['test preparation course']=encodr.fit_transform(x['test preparation course'])
```

```python
x['parental level of education'].unique()
```

```python
encoder2=OrdinalEncoder(categories=[['some high school','high school','some college','associate\'s degree','bachelor\'s degree','master\'s degree']])
```

```python
x[['parental level of education']]=encoder2.fit_transform(x[['parental level of education']])
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,shuffle=True,random_state=42)
```

```python
models={
    'Linear Regression':LinearRegression(),
    'Decision Tree':DecisionTreeRegressor(),
    'KNN':KNeighborsRegressor(),
    'SVR':SVR(),
    'Random Forest':RandomForestRegressor(),
    'AdaBoost':AdaBoostRegressor(),
    'XGBoost':XGBRegressor()
}
```

```python
def evaluate_model(actual,pred):
  mse = mean_squared_error(actual,pred)
  r2 = r2_score(actual,pred)
  return mse,r2
```

```python
models_list=[]
Scores=[]
for i in range(len(list(models))):
  model= list(models.values())[i]
  model.fit(x_train,y_train)
  y_pred= model.predict(x_test)
  mse,r2=evaluate_model(y_test,y_pred)
  print(list(models.keys())[i])

  models_list.append(list(models.keys())[i])
  print('mse:',mse)
  print('r2:',r2)
  Scores.append(r2)


```
