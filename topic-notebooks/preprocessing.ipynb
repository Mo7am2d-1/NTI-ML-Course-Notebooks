# Ordinal Number Encoding 

```python
import numpy as np
import pandas as pd
```

```python
store = pd.DataFrame({
    "size":['xl','l','m','s'],
    "price":[199,290,200,190]
})
```

```python
store
```

```python
mapping={
    'xl':4,
    'l':3,
    'm':2,
    's':1
}
```

```python
#  map
store['size'] = store['size'].map(mapping)
```

```python
store
```

```python
store = pd.DataFrame({
    "size":['xl','l','m','s'],
    "price":[199,290,200,190]
})
```

```python
store
```

```python
#replace
store['size'] = store['size'].replace(mapping)
```

```python
store
```

```python
store = pd.DataFrame({
    "size":['xl','l','m','s'],
    "price":[199,290,200,190]
})
```

```python
# differ between map/replace 
size_map ={
    'm':2,
    'l':3,
    's':4
}
```

```python
store['size'] = store['size'].map(size_map)
```

```python
store
```

```python
store['size'] = store['size'].replace(size_map)
```

```python
store
```

# Handle Missing Value

```python
df = pd.read_csv(r"D:\NTI\machine\session 3 arwa\titanic.csv")
```

Handle Missing values:
    1- Numerical:
        - Values is Enough:
            1- SimpleImputer ---> mean
            2- Fillna() ---> mean()
            3- Drop null values ---> dropna()
        - Values is not enough:
            1- Drop Column
    2- Object:
        - Values in Enough:
            1- Fillna() with most common/frequent value
            2- Drop null values ---> dropna()
        - Values is not Enough:
            1- Drop Column
