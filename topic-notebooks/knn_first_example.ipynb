```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv('/content/weather_forecast.csv')
df.head()
```

```python
df.info()
```

```python
x = df.drop('Play', axis=1)
y = df['Play'].map({'No': 0, 'Yes': 1})
```

```python
x
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
```

```python
x_train_encoded = ohe.fit_transform(x_train)
x_test_encoded = ohe.transform(x_test)
```

# default parameter
KNeighborsClassifier(

    n_neighbors=5,

    weights='uniform',

'uniform': Every neighbor gets exactly one vote. A neighbor 1 inch away has the exact same voting power as a neighbor 100 inches away
'distance': Voting power is weighted by the inverse of their distance. A neighbor right next to the new point gets a massive vote, while a neighbor on the outer edge of the radius gets a tiny fraction of a vote.

    algorithm='auto',

    leaf_size=30,

    p=2,  

Impact: You almost always want p=2 for standard numeric data. p=1 is sometimes better if your data has hundreds of dimensions (features) where diagonal straight lines stop making mathematical sense.

    metric='minkowski',
     
    metric_params=None,
    n_jobs=None
)

```python
from sklearn.neighbors import KNeighborsClassifier

# Initialize the model with customized parameters
model = KNeighborsClassifier(
    n_neighbors=5,       # Look at the 5 closest points (use an odd number)
    weights='distance',  # Closer points get a stronger vote than points further away
    p=2,                 # Use standard straight-line Euclidean distance
    algorithm='auto',    # Let scikit-learn pick the most memory-efficient math structure
    n_jobs=-1            # Use 100% of your computer's CPU cores to speed up calculations
)

model = KNeighborsClassifier(n_neighbors=3)
model.fit(x_train_encoded, y_train)
score = model.score(x_test_encoded, y_test)
print(f"Model score: {score:.2f}\n")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import OneHotEncoder


# 2. Define the range of K values you want to test
# This creates an array of odd numbers from 1 to 29 (e.g., [1, 3, 5, 7... 29])
k_range = list(range(1, 30, 2))

# 3. Create a dictionary that GridSearchCV can read
# The key MUST match the exact parameter name in KNeighborsClassifier
param_grid = {'n_neighbors': k_range}

# 4. Initialize a basic KNN model and the GridSearch
knn = KNeighborsClassifier()

# cv=5 means "Cross-Validation = 5". It will split your training data into 5 chunks,
# train on 4, test on 1, and repeat this for every single K value.
grid = GridSearchCV(knn, param_grid, cv=5, scoring='accuracy', n_jobs=-1)

# 5. Run the massive search!
grid.fit(x_train_encoded, y_train)

# 6. Print the results
print(f"The best K value is: {grid.best_params_['n_neighbors']}")
print(f"Best cross-validated accuracy: {grid.best_score_ * 100:.2f}%")

# 7. You can now use the best model directly to make predictions!
best_model = grid.best_estimator_
# best_model.predict(X_test_encoded)
```

```python
# 1. Test the GridSearch's best model on the unseen test data
grid_final_score = best_model.score(x_test_encoded, y_test)


print(f"Grid Final Score: {grid_final_score}")
```
